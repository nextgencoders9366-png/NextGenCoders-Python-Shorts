```python
import cv2
from matplotlib import pyplot as plt
```

```python
img = cv2.imread("cv_project.jpeg")
plt.figure(figsize=(2.5,2.5))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
plt.figure(figsize=(2.5,2.5))
plt.imshow(gray, cmap="gray")
plt.axis("off")
plt.show()
```

```python
invert = 255 - gray
plt.figure(figsize=(2.5,2.5))
plt.imshow(invert, cmap="gray")
plt.axis("off")
plt.show()
```

```python
blur = cv2.GaussianBlur(invert, (51,51), 0)
plt.figure(figsize=(2.5,2.5))
plt.imshow(blur, cmap="gray")
plt.axis("off")
plt.show()
```

```python
invert_blur = 255 - blur
plt.figure(figsize=(2.5,2.5))
plt.imshow(invert_blur, cmap="gray")
plt.axis("off")
plt.show()
```

```python
sketch = cv2.divide(gray, invert_blur, scale=256)
sketch = cv2.convertScaleAbs(sketch, alpha=1.1, beta=-160)
plt.figure(figsize=(3,3))
plt.imshow(sketch, cmap="gray")
plt.axis("off")
plt.show()
```
